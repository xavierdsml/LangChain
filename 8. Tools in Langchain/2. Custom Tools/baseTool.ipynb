Import Library

```python
from langchain.tools import BaseTool
from typing import Type
from pydantic import BaseModel, Field
```

Pydantic BaseModel

```python
class MultiplicationInput(BaseModel):
  a: int = Field(required=True, description="The first number to add")
  b: int = Field(required=True, description="The second number to add")
```

BaseTools + Function

```python
class MultiplicationTool(BaseTool):
  name: str = "multiply"
  description: str = "Multiply two numbers"
  args_schema: Type[BaseModel] = MultiplicationInput

  def _run(self, a: int, b: int)->int:
    return a*b
```

```python
multiplication_tool = MultiplicationTool()
result = multiplication_tool.invoke({"a":3, "b":4})
```

```python
print(result)
print(multiplication_tool.name)
print(multiplication_tool.description)
print(multiplication_tool.args)
```
