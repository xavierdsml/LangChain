#### Import Lib

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

Pydantic class which is inherit form the BaseModel & in pydantic class we have two attributes which is bascially two integer

```python
class Multiplication(BaseModel):
  a: int = Field(required=True, description="The first number to add")
  b: int = Field(required=True, description="The first number to add")
```

Generate function

```python
def multiplication(a: int, b: int)->int:
  return a*b
```

#### Structure Tool

```python
multiplication_tool = StructuredTool.from_function(
  func=multiplication,
  name='multiply',
  description='Multiplication of two numbers',
  args_schema=Multiplication
)
```

```python
result = multiplication_tool.invoke({"a":3, "b":6})
```

attributes

```python
print(result)
print(multiplication_tool.name)
print(multiplication_tool.description)
print(multiplication_tool.args)
```
