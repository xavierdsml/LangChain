Import Lib

```python
from langchain_core.tools import tool
```

Creating the function

```python
def multiplication(a,b):
  """Multiply two numbers"""
  return a*b

# add type hints
def multiplication(a: int, b: int)->int:
  """Multiply two numbers"""
  return a*b
```

add tool decorator


```python
@tool
def multiplication(a: int, b: int)->int:
  """Multiply two numbers"""
  return a*b
```

Calling the tools

```python
result = multiplication.invoke({"a": 3, "b": 5})
print(result)
```

Three special attribute that the tools should have

```python
print(multiplication.name)
print(multiplication.description)
print(multiplication.args) 
```

The tools that you passed on the LLM , the LLM cannot take as a tools,
the structure that the LLM show as :- args schema based on json

```python
print(multiplication.args_schema.model_json_schema())
```
