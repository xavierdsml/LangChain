Import Library

```python
from langchain_core.tools import tool
```

Tool - 1

```python
@tool
def add(a: int, b: int)->int:
  """Add two numbers"""
  return a+b
```

Tool - 2

```python
@tool
def multiply(a: int, b: int)->int:
  """Multiply two numbers"""
  return a*b
```

Making the ToolKit Function

```python
class MathToolKit:
  def get_tools(self):
    return [add, multiply]
```

```python
toolkit = MathToolKit()
tools = toolkit.get_tools()

for tool in tools:
  print(tool.name, "=>", tool.description)
```
