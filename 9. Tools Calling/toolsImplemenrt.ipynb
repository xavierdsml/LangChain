```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

Tool creation

```python
@tool
def multiply(a:int, b:int)->int:
  """Given two number a and b this tools return's their product"""
  return a*b
```

```python
print(multiply.invoke({'a':3, 'b':4}))
print(multiply.name)
print(multiply.description)
print(multiply.args) # schema
```

### Tool Binding

Create the LLM

```python
llm = HuggingFaceEndpoint(
  repo_id='meta-llama/Llama-4-Scout-17B-16E-Instruct',
  task='text-generation'
)

model = ChatHuggingFace(llm=llm)
```

check the model is working fine or not

```python
model.invoke("hi")
```

Bind...

```python
llm_withTool = model.bind_tools([multiply])
```

```python
llm_withTool.invoke("Hi, who are you ?")
```

### Tool Calling

```python
query = HumanMessage('Can you multiply 3 with 10')
```

```python
messages = [query]
messages
```

```python
result = llm_withTool.invoke(messages)
```

```python
messages.append(result)
messages
```

### Tool Execution

```python
tool_result = multiply.invoke(result.tool_calls[0])
```

```python
messages.append(tool_result)
messages
```

```python
llm_withTool.invoke(messages).content
```
